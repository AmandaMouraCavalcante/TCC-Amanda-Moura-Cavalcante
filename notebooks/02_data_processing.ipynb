```python
%load_ext autoreload
%autoreload 2

import sys
import os
import pandas as pd

# Adiciona a raiz ao path para importar src
project_root = os.path.abspath('..')
if project_root not in sys.path:
    sys.path.append(project_root)

from src.extractors import get_current_year
from src.transformers import clean_ear_data, clean_ena_data, clean_pld_data

RAW_DATA_PATH = os.path.abspath('../data/raw')
year = get_current_year()

# 1. Leitura dos dados brutos
df_ear_raw = pd.read_csv(os.path.join(RAW_DATA_PATH, f"ons_ear_daily_{year}.csv"), sep=";", encoding="utf-8")
df_ena_raw = pd.read_csv(os.path.join(RAW_DATA_PATH, f"ons_ena_daily_{year}.csv"), sep=";", encoding="utf-8")
df_pld_raw = pd.read_csv(os.path.join(RAW_DATA_PATH, f"ccee_pld_hourly_{year}.csv"), sep=";", encoding="utf-8")

# 2. Execução das transformações
df_ear_clean = clean_ear_data(df_ear_raw)
df_ena_clean = clean_ena_data(df_ena_raw)
df_pld_clean = clean_pld_data(df_pld_raw)

print("Transformações concluídas com sucesso!")
```
